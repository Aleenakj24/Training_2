```python
n=int(input())
l=[]
while n>0:
    rem=n%2
    l.append(rem)
    n=n//2
print(l[::-1])#1010
for i in range(len(l)):
    if l[i]==1:
        l[i]=0
    else:
        l[i]=1
print(l[::-1])#0101
p=len(l)-1
d=0
for i in range(len(l)-1,-1,-1):
    d+=l[i]*(2**p)
    p-=1
print(d)
```

```python
n=int(input())
a=list(map(int,input().split()))
k=int(input())

for i in a:
    l=[]
    while i>0:
        rem=i%2
        l.append(rem)
        i=i//2
    for i in range(k):
        if len(l)>0:
            l.pop(0)
    d=0
    p=len(l)-1
    for i in range(len(l)-1,-1,-1):
        d+=l[i]*(2**p)
        p-=1
    print(d,end=" ")

```

```python
n=int(input())
s1=set()
for i in range(n):
    d=int(input())
    s1.add(d)


m=int(input())
s2=set()
for i in range(m):
    d2=int(input())
    s2.add(d2)
s=s1.union(s2)


s3=list(s)
s3.sort()
m=len(s3)//2
if len(s3)%2==0:
    e=(s3[m]+s3[m-1])/2
    print(e)
else:
    print(s3[m])

```
